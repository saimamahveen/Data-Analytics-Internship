### SkillCraft Technology

**Dataset:** Global Superstore Sample

**Objective:**  
Clean and prepare the dataset by identifying and handling missing values,
removing duplicate records, converting data types, and exporting the
cleaned dataset as a CSV file.

## 1. Import Required Libraries

```python
import pandas as pd
```

## 2. Load the Dataset

The Global Superstore dataset is loaded using the Pandas library.

```python
df = pd.read_csv("Global_Superstore_sample.csv")
```

## 4. Identify Missing Values

```python
df.isnull().sum()
```

### Observation

Missing values were identified in the dataset. The missing values were
examined before deciding how they should be handled.

## 5. Handle Missing Values

```python
df["Customer Name"] = df["Customer Name"].fillna("Unknown")
df["Profit"] = df["Profit"].fillna(df["Profit"].median())
```

```python
df.isnull().sum()
```

### Result

Missing values have been handled. The missing Customer Name values were
replaced with "Unknown", while the missing Profit value was replaced
using the median Profit.

## 6. Identify Duplicate Rows

```python
df.duplicated().sum()
```

### Result

Missing values have been handled. The missing Customer Name values were
replaced with "Unknown", while the missing Profit value was replaced
using the median Profit.

## 7. Remove Duplicate Rows

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

### Result

Duplicate rows were identified and removed from the dataset.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

## 8. Convert Data Types

```python
df.dtypes
```

### Result

The Order Date column was converted from a string/object data type to
the datetime data type, making it suitable for date-based analysis.

## 9. Final Data Quality Check

```python
print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

## 10. Export Cleaned Dataset

```python
df.to_csv("Global_Superstore_Cleaned.csv", index=False)
```

```python
print("Cleaned dataset exported successfully.")
```

## Conclusion

The Global Superstore dataset was successfully cleaned and prepared.
Missing values were handled, duplicate records were removed, and the
